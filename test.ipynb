```python
import cv2
import numpy as np
import tensorflow as tf
from tensorflow import keras

IMG_SIZE = 100

# 1) Load the SAME training image file you used (e.g. an 'a' sample)
path = "images\A\A.6a9a3b42-c0ca-11ed-ab21-683e2670f3d6.jpg"

img = cv2.imread(path, cv2.IMREAD_GRAYSCALE)
img = cv2.resize(img, (IMG_SIZE, IMG_SIZE))
img = img.astype("float32") / 255.0
x = np.expand_dims(img, axis=-1)  # shape (100, 100, 1)

# 2) Load model and predict
model = keras.models.load_model("Models/best_model.keras")
pred = model.predict(x[None, ...])  # shape (1, 100, 100, 1)
print(pred.argmax(), pred.max())
```
